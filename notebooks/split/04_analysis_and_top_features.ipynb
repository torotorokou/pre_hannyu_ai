# 04 解析 & 上位特徴量モデル
元ノートの上位10特徴量再学習と可視化部分。

```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error
print('[INFO] libs ready')
```

```python
# 前段で保存された df_weather / feature names を読み込む想定。
# ここでは説明用のプレースホルダ。
try:
    df_weather = pd.read_parquet('df_weather.parquet')
    fn = df_weather.columns[:-1]  # 最終列がターゲット仮定
except Exception:
    print('[WARN] df_weather.parquet が無いのでダミー生成')
    import numpy as np
    fn = [f'f{i}' for i in range(30)]
    data = np.random.randn(200, 31)
    df_weather = pd.DataFrame(data, columns=list(fn)+['target'])
top_10_features = fn[:10]
feature_indices = list(range(10))
X_selected = df_weather.iloc[:, feature_indices]
y = df_weather.iloc[:, -1]
model_top10 = GradientBoostingRegressor(random_state=42).fit(X_selected, y)
pred = model_top10.predict(X_selected)
r2 = r2_score(y, pred); mae = mean_absolute_error(y, pred)
print(f'[RESULT top10] R2={r2:.3f} MAE={mae:.3f}')
imp = model_top10.feature_importances_
imp_df = pd.DataFrame({'feature':X_selected.columns, 'importance':imp}).sort_values('importance', ascending=False)
imp_df.head()
```

```python
# 可視化
plt.figure(figsize=(8,4))
plt.barh(imp_df['feature'][:10], imp_df['importance'][:10])
plt.title('Top10 Feature Importance')
plt.tight_layout()
plt.show()
```
